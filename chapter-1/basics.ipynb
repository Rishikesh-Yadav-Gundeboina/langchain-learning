# Chapter 1: Calling NVIDIA's API three ways

NVIDIA's hosted API (`https://integrate.api.nvidia.com/v1`) is **OpenAI-compatible**, so OpenAI-style clients work by changing `base_url`.

**Setup:** put your NVIDIA key (`nvapi-...`) in `.env` as `OPENAI_API_KEY`.

## 1. OpenAI SDK directly
Plain `openai` client pointed at NVIDIA. Common gotcha: `base_url` must be the API endpoint.

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # loads the key from the local .env file into environment variables

# NVIDIA's endpoint is OpenAI-compatible, so the plain OpenAI SDK works with a different base_url
client = OpenAI(
    base_url="https://integrate.api.nvidia.com/v1",  
    api_key=os.getenv("OPENAI_API_KEY"),  # .env holds the NVIDIA key (nvapi-...) under this name
)

chat_completion = client.chat.completions.create(
    messages=[{"role": "user", "content": "what is an ai agent?"}],
    model="nvidia/nemotron-3.5-lightning-30b-a3b",
)
print(chat_completion.choices[0].message.content)

# Uncomment to list every model ID your key can use (handy when a model name 404s)
# for model in client.models.list().data:
#     print(f"- {model.id}")
```

## 2. LangChain with `ChatOpenAI`
Same endpoint through `langchain-openai`. No NVIDIA-specific package needed.

```python
import os
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    base_url="https://integrate.api.nvidia.com/v1",
    # api_key is omitted: ChatOpenAI falls back to the OPENAI_API_KEY env var (loaded from .env in the cell above)
    model="nvidia/nemotron-3.5-lightning-30b-a3b",
)
print(llm.invoke("what is an ai agent?").content)  # .invoke() returns a message object; .content is the text
```
